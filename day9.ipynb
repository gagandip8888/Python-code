3. Write a function using  *args to calculate the sum, average, maximum, and minimum of any number of integers.  

```python
def calculate(*args):
    total = sum(args)
    average = total / len(args)
    maximum = max(args)
    minimum = min(args)
    return total, average, maximum, minimum
print(calculate(10, 20, 30, 40, 50))
```

5. Write a function that accepts a string and returns the number of uppercase letters, lowercase letters, digits.

```python
def count_characters(text):
    upper = 0
    lower = 0
    digit = 0
    for i in text:
        if i.isupper():
            upper += 1
        elif i.islower():
            lower += 1
        elif i.isdigit():
            digit += 1
    return upper, lower, digit
print(count_characters("Gagan123"))
```

6. Write a function that checks whether a given string is a Palindrome.  

```python
def palindrome(text):
    if text == text[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"
print(palindrome("madam"))
```

7. Write a function that accepts a list of integers and returns a new list containing only even numbers. 

```python
def even_numbers(numbers):
    even = []
    for i in numbers:
        if i % 2 == 0:
            even.append(i)
    return even
list1 = [10, 15, 22, 33, 40, 51]

print(even_numbers(list1))
```

9. Write a function that accepts a dictionary of students and their marks, then returns the name of the topper and the highest marks.  

```python
students = {
    "Gagan": 85,
    "Omkar": 92,
    "Sarthak": 78,
    "Prathmesh": 95
}
def topper(marks):
    highest = max(marks.values())
    for name in marks:
        if marks[name] == highest:
            return name, highest

print(topper(students))
```

10. Write a function that accepts a sentence and returns the frequency of each word using a dictionary. 

```python
def word_frequency(sentence):
    words = sentence.split()
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    return frequency
print(word_frequency("hello how are you"))
```

11. Write a function that accepts a list of integers and returns:  • Sum of all numbers  • Count of even numbers  • Count of odd numbers  

```python
def numbers(list1):
    total = 0
    even = 0
    odd = 0
    for i in list1:
        total += i
        if i % 2 == 0:
            even += 1
        else:
            odd += 1
    return total, even, odd
list1 = [10, 15, 20, 25, 30]
print(numbers(list1))
```
