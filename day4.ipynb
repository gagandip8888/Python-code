1. Write a program to count the number of vowels and consonants in a given string. 

```python
string = input("Enter a string: ")
vowel = 0
consonant = 0
for ch in string:
    if ch.lower() in "aeiou":
        vowel += 1
    else:
        consonant += 1
print( 'count of vowel=',vowel)
print('count of consonant=',consonant)
```

2. Write a program to check whether a given string is a Palindrome. - - Input: madam Output : Palindrome 

```python
palidrome=input('enter palidrome world')
if palidrome==palidrome [::-1]:
    print('palidrome')
else:
    print('no palidrome')
```

5. Write a program to remove all duplicate characters from a string while maintaining their original order. - - Input : programming Output : progamin 

```python
program=input('enter a duplicate value')
res=''
for ch in program:
    if ch not in res:
        res=res+ch
        print('original value:',res)
```

6. Write a program to find the longest word in a sentence. - Input: Python is an amazing programming language  - Output: programming 

```python
long=" Python is an amazing programming language " 
words= long.split()
longest=''
for word in words:
    if len(word)>len (longest):
        longest=word
        print('long word:',longest)
    
```

7. Write a program to capitalize the first letter of every word without using the title() method. - - Input:    welcome to python Output:   Welcome To Python 

```python
string =input('Enter a sentence ')
words=string.split()
result=''
for word in words:
    result =result+word[0].upper()+ word[1:]+''
    print (result)
```
