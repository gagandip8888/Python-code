1. Write a program to find the union of two sets without using the union() method.

```python
sets1={10,20,30}
sets2={20,30,40,50}
sets3=set()
for i in sets1:
    sets3.add(i)
for j in sets2:
    sets3.add(j)
    
print(sets3)
    
```

3. Write a program to find all common elements between two lists using sets. 

```python
list1=[10,20,30,40]
list2=[10,20,30]
list3=set()

for i in list1:
    if i in list2:
        list3.add(i)
print(list3)
```

5. Write a program to find the missing numbers between 1 to N using sets. - - - -  - - Input: N = 10 Numbers = {1,2,3,5,7,8,10} Output: {4,6,9} 

```python
n=10
num={1,2,3,5,7,8,10}
num1=set()
for i in range(1,n+1):
    if i not in num:
     num1.add(i)
print (num1)
```

6. Write a program to create a new set containing only the even numbers from a given set. 

```python
num = {10,15,20,25,30}
num1=set()

for i in num:
     if i%2==0:
         num1.add(i)
         
print(num1)
         
    
```

7. Write a program to merge three sets into a single set containing only unique elements. 

```python
set1={10,20,30,40,50}
set2={20,30,40}
set3={10,60 ,70}
set4=set()

for i in set1:
    set4.add(i)
    for j in set2:
        set4.add(j)
        for k in set3:
            set4.add(k)
            
print(set4)


            
            
```
