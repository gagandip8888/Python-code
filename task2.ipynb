2. Write a Python program using nested loops to find and print all prime numbers between 1 and 100.

```python
for i in range(2,101):
    count=0
    
    for j in range(1,i+1):
        if i%j==0:
            count=count +1 
    if count==2:
        print(i)
```

9. Write a Python program to count the frequency of each word in a sentence and store the result in a dictionary.  

```python
sentence = "apple banana apple mango banana apple"
word1 = {}

for word in sentence.split():
    if word in word1:
        word1[word] += 1
    else:
        word1[word] = 1

print(word1)
    
```

11. Create a function calculator(a, b, operation) that performs addition, subtraction, multiplication, and division based on the given operation

```python
def calculator(a,b,operation):
    if operation=="add":
        print(a+b)
        
    elif operation=="sub":
        print(a-b)
        
    elif operation=="mul":
        print(a*b)
        
    elif operation=="div":
        print(a/b)
        
    else:
        print("no number")
        
        
calculator(10,5,"add")
calculator(20,10,"sub")
calculator(20,10,"mul")
calculator(100,2,"div")

        
```

12. Create a function that accepts a list of numbers and returns the largest number, smallest number, sum, and average. 

```python
def calculate(numbers):
    largest=max(numbers)
    smallest=min(numbers)
    total=sum(numbers)
    average=total/len(numbers)
    
    return largest,smallest,total,average

numbers=[10,20,30,40,50]
result=calculate(numbers)
print(result[0])
print(result[1])
print(result[2])
print(result[3])
```

13. Create a BankAccount class with account holder, account number, and balance. Add methods for deposit, withdrawal, and displaying balance. Prevent withdrawal when the balance is insufficient. 

```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance=0):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display_balance(self):
        print("Current balance:", self.balance)
        
account = BankAccount("Rahul", 12345, 5000)
account.deposit(1000)
account.withdraw(2000)
account.display_balance()

```

14. Create an Employee class with employee name, ID, department, and salary. Add methods to display details, increase salary by a given percentage, and check whether the salary is above ₹50,000. 

```python
class Employee:
    def __init__(self, name, id, department, salary):
        self.name = name
        self.id = id
        self.department = department
        self.salary = salary

    def display_details(self):
        print("Name:", self.name)
        print("ID:", self.id)
        print("Department:", self.department)
        print("Salary:", self.salary)

    def increase_salary(self, percentage):
        self.salary += self.salary * percentage / 100

    def is_above_50000(self):
        return self.salary > 50000


employee = Employee("Rahul", 101, "IT", 50000)
employee.display_details()
employee.increase_salary(10)
print("Above ₹50,000:", employee.is_above_50000())
```

15. Create a Vehicle class with brand and speed. Create Car and Bike child classes with additional attributes. Use inheritance and method overriding to display their complete details.  

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def display_details(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed)

class Car(Vehicle):
    def __init__(self, brand, speed, doors):
        super().__init__(brand, speed)
        self.doors = doors

    def display_details(self):
        super().display_details()
        print("Number of doors:", self.doors)


class Bike(Vehicle):
    def __init__(self, brand, speed, engine_cc):
        super().__init__(brand, speed)
        self.engine_cc = engine_cc

    def display_details(self):
        super().display_details()
        print("Engine capacity:",self.engine_cc, "cc")


car = Car("Toyota", 180, 4)
bike = Bike("Yamaha", 120, 150)

car.display_details()
print()
bike.display_details()
    
```
