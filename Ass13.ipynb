6. Write a program to read a file and create another file containing only the lines that contain the word "Python".  

```python
with open("studntdata.txt", "r") as file:
    with open("studntdata_python.txt", "w") as newfile:
        for line in file:
            if "python" in line.lower():
                newfile.write(line)
```

8. Write a program to read a file and find the longest word present in it.

```python
with open("studntdata.txt","r")as file:
    data=file.read()
    
words=data.split()
longest=""
for i in words:
     if len(i) > len(longest):
        longest=i

print(longest)
        
```

9. Write a program to read a file containing numbers and calculate their sum, average, maximum, and minimum.  

```python
with open("calculate.txt", "r") as file:
    data = file.read()
numbers = []
for i in data.split():
    numbers.append(int(i))
print("Sum of numbers:", sum(numbers))
print("Average of numbers:", sum(numbers) / len(numbers))
print("Maximum of numbers:", max(numbers))
print("Minimum of numbers:", min(numbers))
```
