2. Write a program to count the number of even and odd numbers in a tuple

```python
num = (1,23,45,6,78,90,8)
Even = 0
odd = 0
for i in num:
    if i % 2 == 0:
        Even = Even + 1
    else:
        odd = odd + 1

print(Even,'even ')
print(odd,'odd')
```

3. Write a program to find all common elements between two tuples. 


```python
tup1=(20,30,40,50)
tup2=(20,30,70)
tup3=[]

for i in tup1:
    if i in tup2:
        tup3.append(i)

print(tuple(tup3))
```

4. Write a program to create a new tuple containing only positive numbers from a given tuple. 

```python
tp1=(10,-20,4,5,-66)
tp2=[]
for i in tp1:
    if i >0:
        tp2.append(i)
        
print(tuple(tp2))
        
```

6. Write a program to separate all even and odd elements of a tuple into two different tuples. - - -  - - - Input: (10,15,20,25,30) Output: Even = (10,20,30) Odd = (15,25) 

```python
num=(10,15,20,25,30)
Even=[]
odd=[]

for i in num:
    if i%2==0:
        Even.append(i)
    else:
        odd.append(i)
        
print("Even:",tuple(Even))
print("Odd:",tuple(odd))
```

8. Write a program to find the union of two tuples (unique elements only). 

```python
tuple1=(10,20,30)
tuple2=(20,30,40,50)
tuple3=set()
for i in tuple1:
    tuple3.add(i)
for j in tuple2:
    tuple3.add(j)
    
        
print(tuple(tuple3))


```

9. Write a program to convert a nested tuple like: Input - ((1,2), (3,4), (5,6)) into: output -(1,2,3,4,5,6) 

```python
num=((1,2),(3, 4),(5,6))
num2=[]
for i in num:
    for j in i:
        num2.append(j)
print(tuple(num2))
    
```
