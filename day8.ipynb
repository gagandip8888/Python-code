1. Write a program to count the frequency of each character in a string using a dictionary. - - -  - - Input: "programming" Output: {'p':1, 'r':2, 'o':1, ...} 

```python
str1="programming"
str2={}

for i in str1:
    if i in str2:
        str2[i] += 1
    else:
        str2[i]= 1

print(str2)
            
```

2. Write a program to count the frequency of each element in a list using a dictionary. - - -  - - [10,20,10,30,20,10] Output: {10:3, 20:2, 30:1} 

```python
num=[10,20,10,30,20,10]
num1={}

for i in num:
    if i in num1:
        num1[i]+=1
    else:
        num1[i]=1
print(num1)
```

3. Write a program to find the key having the maximum value in a dictionary. - - -  - - {'A':85, 'B':92, 'C':78} Output: B 

```python
value={'A':85,'B':92,'C':78}
maximum=0
for i in value:
    if value[i]>maximum:
        maximum=value[i]
        maximum_key=i
print(maximum_key)
```

4. Write a program to find the key having the minimum value in a dictionary

```python
value={'A':88,'B':37,'C':44}
minimum = max(value.values())
for i in value:
    if value[i]< minimum :
        minimum=value[i]
        minimum_key=i
print(minimum_key)
```

6. Write a program to remove duplicate values from a dictionary. - - -  - - {'a':10, 'b':20, 'c':10, 'd':30} Output: {'a':10, 'b':20, 'd':30}

```python
dict1={'a':10,'b':20,'c':10,'d':30}
dict2=set()
dict3={}


for i in dict1:
    if dict1[i]not in dict2:
        dict2.add(dict1[i])
        dict3[i]=dict1[i]
print(dict3)
```

8. Write a program to find the sum of all values in a dictionary. - {'Math':85, 'Science':90, 'English':80} -  - - Output: 255 

```python
marks={'Math':85,'Science':90,'English':80}
total = 0
for i in marks.values():
    total += i
print(total)
```

9. Write a program to create a dictionary from two lists: one containing keys and the other containing values. - - Keys = ['A', 'B', 'C'] Values = [10, 20, 30] 

```python
keys=['A','B','C']
values=[10,20,30]
dict1={}
for i in range (len(keys)):
  dict1[keys[i]] = values[i]
print(dict1)
    
```
