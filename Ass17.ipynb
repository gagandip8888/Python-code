```python
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance
        
    def deposite(self,ammount):
        if ammount>0:
            self.__balance+=ammount
            print("Amoount is deposite")
            
        else:
            print("dont allow negative deposite")

    def withdraw(self,ammount):
        if ammount>self.__balance:
            print("no balance")
        else:
            self.__balance-=ammount
            print("ammount is withdraw")
            
            
    def get_balance(self):
        return self.__balance
    
    
account1=BankAccount(1000)
account1.deposite(100)
account1.withdraw(500)

print("your balance is:",account1.get_balance())
```

9. Payment System — Method Overriding + Polymorphism Create a parent class Payment. Create child classes: Payment ├── UPI ├── CreditCard └── NetBanking • Create pay() method in parent.  • Override pay() in each child.  • Create a function make_payment(payment). 

```python
class payment:
    def pay(self):
        print("payment processing")

class UPI(payment):
    def pay(self):
        print("payment done using upi")
    
class CreditCard(payment):
    def pay(self):
        print('payment done using credit card')
class NetBanking(payment):
    def pay(self):
        print("payment done using Net Banking")
        

def make_payment(payment):
    payment.pay()
    
upi1 = UPI()
card1 = CreditCard()
net1 = NetBanking()

make_payment(upi1)
make_payment(card1)
make_payment(net1)

```

10. Calculator — Method Overloading Create a Calculator class. Create an add() method that should support: add(10, 20) add(10, 20, 30) add(10, 20, 30, 40) • Use *args or default arguments.  • Do not create multiple add() methods.  • Similarly implement multiply() using *args.  

```python
class Calculator:
    def add(self, *args):
        return sum(args)
    
    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        return result

calc = Calculator()
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
print(calc.multiply(10, 20))
print(calc.multiply(10, 20, 30))
```

Library Management — Aggregation + Encapsulation Create: Book Library • Book should contain title, author, and price.  • Make price private.  • Create getter/setter for price.  • Library should contain multiple Book objects.  • Add books to the library.  • Display all books.  • Search for a book by title.  

```python
class Book:
    def __init__(self,title,author,price):
        self.title=title
        self.author=author
        self.__price=price
        
    def get_price(self):
        return self.__price
    
    def set_price(self,price):
        self.__price=price
        
class Library:
    def __init__(self):
        self.books = []
        
    def add_book(self, book):
        self.books.append(book)
        
    def display_books(self):
        for book in self.books:
            print("Title:",book.title)
            print("Author:",book.author)
            print("Price:",book.get_price())
            print()
            
    def search_book(self, title):
        for book in self.books:
            if book.title == title:
                print("Book Found")
                print("Title:", book.title)
                print("Author:", book.author)
                print("Price:", book.get_price())
                return
        print("Book Not Found")
        
book1 = Book("Python", "sarthak", 500)
book2 = Book("Java", "swapnil", 600)
book3 = Book("C++", "prathmesh", 700)

library1 = Library()

library1.add_book(book1)
library1.add_book(book2)
library1.add_book(book3)

print("All Books:")
library1.display_books()

print("Search Result:")
library1.search_book("Python")

```

12. E-Commerce Product — Encapsulation + Static Method Create a Product class. • Private variable __price.  • Instance variables: name, quantity.  • Getter/setter for price.  • Instance method total_price().  • Static method calculate_gst(price) that calculates GST.  • Display the final product price including GST.

```python
class Product:
    def __init__(self, name, quantity, price):
        self.name = name
        self.quantity = quantity
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price >= 0:
            self.__price = price
        else:
            print("Price cannot be negative")

    def total_price(self):
        return self.quantity * self.__price

    @staticmethod
    def calculate_gst(price):
        gst_rate = 0.18
        return price * gst_rate

    def display(self):
        base_total = self.total_price()
        gst = self.calculate_gst(base_total)
        final_total = base_total + gst

        print("Product Name:", self.name)
        print("Quantity:", self.quantity)
        print("Price per unit:", self.__price)
        print("Total Price:", base_total)
        print("GST:", gst)
        print("Final Price Including GST:", final_total)
        print()

p1 = Product("Laptop", 2, 50000)
p1.display()
```

13. Hospital Management — Aggregation + Inheritance Create: Person ├── Doctor └── Patient and an Address class. • Doctor and Patient should inherit from Person.  • Both should contain an Address object.  • Add different attributes for Doctor and Patient.  • Override display() in both child classes.  • Use super() where required.  This combines inheritance + aggregation + overriding. 

```python
class Address:
    def __init__(self, city, state, pincode):
        self.city = city
        self.state = state
        self.pincode = pincode

    def display(self):
        print(f"City: {self.city}")
        print(f"State: {self.state}")
        print(f"Pincode: {self.pincode}")


class Person:
    def __init__(self, name, age, address):
        self.name = name
        self.age = age
        self.address = address

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Address:")
        self.address.display()


class Doctor(Person):
    def __init__(self, name, age, address, specialization, experience):
        super().__init__(name, age, address)
        self.specialization = specialization
        self.experience = experience

    def display(self):
        print("Doctor Details")
        super().display()
        print("Specialization:", self.specialization)
        print("Experience:", self.experience, "years")
        print()


class Patient(Person):
    def __init__(self, name, age, address, disease, room_no):
        super().__init__(name, age, address)
        self.disease = disease
        self.room_no = room_no

    def display(self):
        print("Patient Details")
        super().display()
        print("Disease:", self.disease)
        print("Room No:", self.room_no)
        print()

addr1 = Address("Pune", "Maharashtra", 411001)
addr2 = Address("Nagpur", "Maharashtra", 440001)


doc = Doctor("Dr. Ramesh", 38, addr1, "Cardiologist", 12)
pat = Patient("Amit", 28, addr2, "Fever", "B-204")

doc.display()
pat.display()
```
